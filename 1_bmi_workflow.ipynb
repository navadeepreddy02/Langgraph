```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
#define state
class BMIState(TypedDict):
    wight : float
    height : float
    bmi : float
    category : str
```

```python
def calculate_bmi(state:BMIState) -> BMIState:
    weight = state['wight']
    height = state['height']

    bmi = weight/(height**2)

    state['bmi'] = round(bmi,2)

    return state
```

```python
def catogery_all(state:BMIState) -> BMIState:
    bmi = state['bmi']

    if bmi < 18.5:
        state["category"] = "Underweight"
    elif 18.5 <= bmi < 25:
        state["category"] = "Normal"
    elif 25 <= bmi < 30:
        state["category"] = "Overweight"
    else:
        state["category"] = "Obese"

    return state
```

```python
#define your graph
graph = StateGraph(BMIState)

#Add nodes
graph.add_node('calculate_bmi',calculate_bmi)
graph.add_node("catogery_all",catogery_all)

#add edges
graph.add_edge(START,"calculate_bmi")
graph.add_edge("calculate_bmi","catogery_all")
graph.add_edge("catogery_all",END)


#compile the graph (check the mistakes)
workflow = graph.compile()

#execute the graph
intital_state = {"wight":50,"height":5.5}

final_state = workflow.invoke(intital_state)

print(final_state)
```
